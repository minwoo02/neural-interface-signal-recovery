# 02. Baseline Signal Model

이번 단계에서는 신경 기록 인터페이스의 손상이나 회복을
도입하기 전에 정상 상태의 최소 신호 모델을 정의한다.

관측 신호를

$$
y(s)=x(s)+n(s)
$$

로 정의한다.

여기서

- $x(s)$: 기록하고 싶은 기준 입력 신호
- $n(s)$: 입력에 더해지는 잡음
- $y(s)$: 실제 관측 신호

를 의미한다.

이 단계에서는 건강한 인터페이스를 가정하여

$$
h=1
$$

로 두고, 별도의 증폭 보정도 사용하지 않아

$$
g=1
$$

로 둔다.

따라서 이 단계의 목적은 실제 신경 신호를 정밀하게
재현하는 것이 아니라, 이후 손상과 회복의 효과를
비교하기 위한 기준 상태(baseline)를 만드는 것이다.


```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
SEED = 42
rng = np.random.default_rng(SEED)
```

```python
#기준입력 x(s) 만들기
n_samples = 512
s = np.arange(n_samples) / n_samples
x = (
    np.sin(2 * np.pi * 7 * s) + 0.35 * np.sin(2 * np.pi * 19 * s)
)
```

```python
x = x - x.mean() #평균 0으로(DC offset 없이)
```

```python
x = x / np.sqrt(np.mean(x**2)) #RMS 1로 정규화

assert np.isclose(
    np.mean(x), 0.0, atol=1e-14
)

assert np.isclose(
    np.sqrt(np.mean(x**2)), 1.0

)

print(f"Input mean = {np.mean(x):.3e}")
print(f"Input RMS = {np.sqrt(np.mean(x**2)):.3f}")

plt.figure(figsize=(9,4))
plt.plot(s, x)
plt.xlabel("Normalized sample position")
plt.ylabel("Normalized amplitude")
plt.title("Baseline synthetic input x(s)")
plt.tight_layout()
plt.show()
```

```python
#잡음 n(s) 추가
sigma = 0.2

noise = rng.normal(
    loc=0.0, scale=sigma, size=n_samples
)
```

```python
#실제 관측 신호 만들기
y = x + noise

fig, axes = plt.subplots(
    3, 1, figsize=(10, 8), sharex=True
)

#x, n, y 세 신호그래프 같이보기
axes[0].plot(s, x)
axes[0].set_ylabel("Amplitude")
axes[0].set_title("Clean signal x(s)")
axes[1].plot(s, noise)
axes[1].set_ylabel("Amplitude")
axes[1].set_title("Noise n(s)")
axes[2].plot(s, y)
axes[2].set_ylabel("Amplitude")
axes[2].set_xlabel("Normalized sample position")
axes[2].set_title("Observed signal y(s) = x(s) + n(s)")
plt.tight_layout()
plt.show()
```

```python
#겹쳐보기
plt.figure(figsize=(10, 4))

plt.plot(
    s[:150],
    x[:150],
    "k--",
    label="Clean x(s)"
)

plt.plot(
    s[:150],
    y[:150],
    label="Observed y(s)"
)

plt.xlabel("Normalized sample position")
plt.ylabel("Normalized amplitude")
plt.title("Baseline recording")

plt.legend()
plt.tight_layout()
plt.show()
```

```python
#RMS 계산
x_rms = np.sqrt(np.mean(x**2))
noise_rms = np.sqrt(np.mean(noise**2))
y_rms = np.sqrt(np.mean(y**2))

print(f"Signal RMS   = {x_rms:.3f}")
print(f"Noise RMS    = {noise_rms:.3f}")
print(f"Observed RMS = {y_rms:.3f}")
```

```python
#SNR
snr_db = 20 * np.log10(
    x_rms / noise_rms
)

print(f"Baseline SNR = {snr_db:.2f} dB")
```

```python
assert np.isclose(
    np.mean(x),
    0.0,
    atol=1e-14
)

assert np.isclose(
    np.sqrt(np.mean(x**2)),
    1.0
)

assert len(x) == n_samples
assert len(noise) == n_samples
assert len(y) == n_samples

assert np.allclose(
    y,
    x + noise
)

print("Baseline signal model validation passed.")
```
